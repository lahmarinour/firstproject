```python
import pandas as pd

# 1. Chargement des données traitées
df = pd.read_csv("stocktwits_sentiment_processed.csv")
print(f"📂 Dataset chargé : {len(df)} messages à agréger.")


# 2. Fonction de regroupement et de calcul
def compute_sentiment_summary(df_input):
    """Calcule le volume de messages, la moyenne VADER et la répartition par ticker."""
    summary = (
        df_input.groupby("ticker")
        .agg(
            total_msg=("cleaned_text", "count"),
            vader_mean=("vader_compound", "mean"),
            vader_median=("vader_compound", "median"),
            bullish_pct=(
                "vader_sentiment",
                lambda x: (x == "Positive").mean() * 100,
            ),
            bearish_pct=(
                "vader_sentiment",
                lambda x: (x == "Negative").mean() * 100,
            ),
        )
        .reset_index()
    )

    # Tri par score moyen décroissant (actions les plus positives en premier)
    summary = summary.sort_values(by="vader_mean", ascending=False)
    return summary


# 3. Exécution du calcul
df_summary = compute_sentiment_summary(df)

# 4. Aperçu des résultats dans la console
print("\n--- Tableau récapitulatif par Action ---")
print(df_summary.to_string(index=False))

# 5. Sauvegarde du tableau agrégé
df_summary.to_csv("stocktwits_sentiment_summary.csv", index=False)
print(
    "\n✅ Enregistré dans 'stocktwits_sentiment_summary.csv' !"
)
```
