```python
#first draft lol
import re
import pandas as pd


# 1. Charger les données brutes générées à l'étape précédente
df = pd.read_csv("stocktwits_sentiment_raw.csv")
print(f"📊 Nombre de lignes avant nettoyage : {len(df)}")


# 2. Fonction de nettoyage de texte (NLP Preprocessing)
def clean_text(text):
    if not isinstance(text, str):
        return ""

    # Supprimer les URLs (http, https, www)
    text = re.sub(r"https?://\S+|www\.\S+", "", text)

    # Supprimer les mentions StockTwits/Twitter (@username)
    text = re.sub(r"@\w+", "", text)

    # Supprimer les espaces multiples et retours à la ligne
    text = re.sub(r"\s+", " ", text).strip()

    return text


# 3. Appliquer le nettoyage
df["cleaned_text"] = df["text"].apply(clean_text)

# Supprimer les lignes où le texte est devenu vide après nettoyage
df = df[df["cleaned_text"] != ""].copy()

# 4. Harmoniser les dates en format Datetime Python
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

# 5. Aperçu des données nettoyées
print(f"✅ Nombre de lignes après nettoyage : {len(df)}")
print(df[["ticker", "created_at", "sentiment_tag", "cleaned_text"]].head())

# 6. Sauvegarder la version propre des données
df.to_csv("stocktwits_sentiment_cleaned.csv", index=False)
print("💾 Fichier 'stocktwits_sentiment_cleaned.csv' sauvegardé !")
"""
import re
import pandas as pd
from tqdm import tqdm

# 1. Charger les données brutes
print("📂 Chargement du fichier 'stocktwits_sentiment_raw.csv'...")
df = pd.read_csv("stocktwits_sentiment_raw.csv")
print(f"📊 Lignes brutes chargées : {len(df)}")


# 2. Fonction de nettoyage texte
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = re.sub(r"https?://\S+|www\.\S+", "", text)  # Supprimer URLs
    text = re.sub(r"@\w+", "", text)  # Supprimer mentions @user
    text = re.sub(r"\s+", " ", text).strip()  # Normaliser espaces
    return text


# 3. Application du nettoyage avec suivi tqdm
tqdm.pandas(desc="🧹 Nettoyage du texte")
df["cleaned_text"] = df["text"].progress_apply(clean_text)

# Supprimer les lignes devenues vides
df = df[df["cleaned_text"] != ""].copy()

# 4. Conversion et harmonisation des dates
print("⏳ Harmonisation des formats de date...")
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

# 5. Aperçu rapide
print("\n--- Aperçu des 5 premières lignes nettoyées ---")
print(df[["ticker", "created_at", "sentiment_tag", "cleaned_text"]].head())

# 6. Sauvegarde du fichier propre
df.to_csv("stocktwits_sentiment_cleaned.csv", index=False)
print(
    f"\n✅ Nettoyage terminé ! {len(df)} lignes sauvegardées dans 'stocktwits_sentiment_cleaned.csv'."
)"""
```
