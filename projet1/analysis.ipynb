```python
import nltk
import pandas as pd
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Téléchargement du lexique VADER (à faire une seule fois)
nltk.download("vader_lexicon")

# 1. Charger les données nettoyées
df = pd.read_csv("stocktwits_sentiment_cleaned.csv")
print(f"📂 Dataset chargé : {len(df)} lignes.")

# 2. Initialiser l'analyseur VADER
analyzer = SentimentIntensityAnalyzer()


# 3. Fonction pour calculer les scores
def get_vader_score(text):
    if not isinstance(text, str):
        return 0.0
    # VADER renvoie un dictionnaire avec : pos, neg, neu, compound
    return analyzer.polarity_scores(text)["compound"]


# 4. Appliquer VADER sur le texte nettoyé
print("🧠 Calcul des scores de sentiment avec VADER...")
df["vader_compound"] = df["cleaned_text"].apply(get_vader_score)


# 5. Categoriser le score en 3 classes (Positive, Negative, Neutral)
def categorize_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"


df["vader_sentiment"] = df["vader_compound"].apply(categorize_sentiment)

# 6. Aperçu des résultats
print("\n--- Échantillon des scores calculés ---")
print(
    df[
        [
            "ticker",
            "cleaned_text",
            "sentiment_tag",
            "vader_compound",
            "vader_sentiment",
        ]
    ].head()
)

# 7. Sauvegarde finale du dataset enrichi
df.to_csv("stocktwits_sentiment_processed.csv", index=False)
print(
    "\n✅ Terminé ! Résultats sauvegardés dans 'stocktwits_sentiment_processed.csv'."
)
```
